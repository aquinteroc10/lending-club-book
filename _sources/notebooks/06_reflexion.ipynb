## Análisis de modelos
PySpark fue más lento porque el entorno de ejecución no corresponde a su caso de uso diseñado. El overhead de coordinación distribuida (serialización JVM/Python, planificación de DAGs, gestión de particiones) supera los beneficios de la paralelización cuando se opera en modo local con 2 cores. En un clúster real con 10+ nodos, los 27 jobs del CrossValidator correrían en paralelo real y PySpark superaría a scikit-learn tanto en tiempo como en escalabilidad.
## ¿Qué aporta LIME y cuáles son sus limitaciones en entornos distribuidos?
LIME construye, alrededor de una predicción específica, un modelo lineal simple que aproxima el comportamiento local del modelo complejo. Para este proyecto, si el modelo rechaza un préstamo, LIME puede mostrar que *dti* alto y *fico_range_high* bajo fueron los factores determinantes para ese solicitante específico.
### Limitaciones
- LIME requiere el modelo en el driver: ara explicar una predicción, LIME necesita llamar al modelo cientos de veces sobre perturbaciones de la instancia original.
- No opera sobre DataFrames distribuidos:  LIME trabaja sobre una fila a la vez convertida a NumPy/Pandas.
- Las explicaciones no son globales: LIME explica instancias individuales.
